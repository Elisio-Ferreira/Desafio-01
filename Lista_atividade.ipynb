```python
#1- Faça um programa que calcule a média de três números inseridos pelo usuário.

number1 = float(input("Digite o primeiro número: "))
number2 = float(input("Digite o segundo número: "))
number3 = float(input("Digite o terceiro número: "))

result = (number1 + number2 + number3) / 3

print("A média dos números é:", result)

```

```python
#2- Crie um programa que determine se um número inserido pelo usuário é par ou ímpar.


while True:
  number = int(input(" Digite um número: "))

  if number % 2 == 0:
    print("\n ------ O número é par. ------")
  else:
    print("\n ------ O número é ímpar. ------")
```

```python
#3- Escreva um programa que solicite um número ao usuário e imprima todos os números pares de 0 atéesse número.

while True:
  number = int(input("Digite um número: "))


  for pares in range(0,number + 1):
    if pares % 2 == 0:
      print(pares)
```

```python
#4- Crie um programa que leia uma lista de números e exiba o maior e o menor valor da lista.

mylister = [10,17,2,24,9,1,3,15,11,18,22,13,12];

print(" ----- O maior valor é: ", max(mylister),"-----")
print("\n ----- O menor valor é: ", min(mylister),"-----")
```

```python
#5- Faça um programa que leia uma lista de números e retorne a média dos números pares.

mylister = [10,17,2,24,9,1,3,15,11,18,22,13,12];

pares = []
for i in mylister:
  if i % 2 == 0:
    pares.append(i)

media = sum(pares) / len(pares)
print(media)

```

```python
#7- Crie um programa que imprima a sequência de Fibonacci até um valor inserido pelo usuário.

while True:
  valor = int(input("Digite um valor: "))

  valorInicial = 0
  valorFinal = 1

  for i in range(valor):

    valorInicial, valorFinal = valorFinal, valorInicial + valorFinal
    print(valorInicial)

```

```python
#8- Faça um programa que determine se um número é primo ou não.

while True:
    valor = int(input("Digite um valor: "))

    if valor <= 1:
        print("Não é primo")

    else:
        for i in range(2, valor):
            if valor % i == 0:
                print("Não é primo")
                break
        else:
            print("É primo")
```
